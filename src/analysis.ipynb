```python
import pandas as pd

# carregar dataset
df = pd.read_csv("../data/goodreads_books_dataset.csv")
```

```python
df.head()
```

```python
# remover colunas irrelevantes (ajusta conforme dataset)
df = df[['title', 'author', 'rating', 'description', 'genres', 'pages']]

# remover linhas sem descrição
df = df.dropna(subset=['description'])

# remover duplicados
df = df.drop_duplicates(subset=['title'])

# reset index
df = df.reset_index(drop=True)

print(df.shape)
```

```python
df['content'] = df['title'] + " " + df['description']
```

```python
df.head()
```

```python
df.to_csv("../data/clean_books.csv", index=False)
```
